```python
#Device 2
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ==== Input files ====
file_05 = "/content/drive/MyDrive/tcad/jfet2/Id_Vg_Drain_0.5V.csv"
file_1  = "/content/drive/MyDrive/tcad/jfet2/Id_Vg_Drain_1V.csv"

# ==== Load data ====
df05 = pd.read_csv(file_05)
df1  = pd.read_csv(file_1)

# Auto-detect column names
vg_col = df05.columns[0]   # VGS
id_col = df05.columns[1]   # IDS

# --- Handle duplicate VGS values by averaging IDS for each VGS ---
df05u = df05.groupby(vg_col, as_index=False)[id_col].mean().sort_values(by=vg_col)
df1u  = df1.groupby(vg_col,  as_index=False)[id_col].mean().sort_values(by=vg_col)

Vg_05 = df05u[vg_col].values
Id_05 = df05u[id_col].values

Vg_1  = df1u[vg_col].values
Id_1  = df1u[id_col].values

# ==== Compute gm = dId/dVg ====
gm_05 = np.gradient(Id_05, Vg_05)
gm_1  = np.gradient(Id_1,  Vg_1)

# ==== Find gm_max and corresponding VGS ====
idx_05 = np.argmax(gm_05)
idx_1  = np.argmax(gm_1)

gm_max_05 = gm_05[idx_05]
vg_at_gm_max_05 = Vg_05[idx_05]

gm_max_1 = gm_1[idx_1]
vg_at_gm_max_1 = Vg_1[idx_1]

print("===== gm_max Results =====")
print(f"VDS = 0.5 V: gm_max = {gm_max_05:.6e} A/V  ({gm_max_05*1e3:.3f} mS) at VGS = {vg_at_gm_max_05:.3f} V")
print(f"VDS = 1.0 V: gm_max = {gm_max_1:.6e} A/V  ({gm_max_1*1e3:.3f} mS) at VGS = {vg_at_gm_max_1:.3f} V")

# ==== Plot gm vs VGS ====
plt.figure(figsize=(7,5))
plt.plot(Vg_05, gm_05, label="gm @ VDS = 0.5 V")
plt.plot(Vg_1,  gm_1,  label="gm @ VDS = 1.0 V")
plt.xlabel("VGS (V)")
plt.ylabel("gm (A/V)")
plt.title("Transconductance gm vs VGS")
plt.grid(True)
plt.legend()
plt.show()
```
